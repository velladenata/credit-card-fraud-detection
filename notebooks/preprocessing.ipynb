#### **Undersampling** (untuk balancing, mengurangi jumlah sampel kelas mayoritas)

```python
import imblearn
from imblearn.under_sampling import RandomUnderSampler

undersample = RandomUnderSampler(sampling_strategy=0.5)
```

```python
X = data_validated.drop(columns=['Class'])
Y = data_validated['Class']

# undersample
X_undersample, Y_undersample = undersample.fit_resample(X, Y)
```

```python
test = pd.DataFrame(Y_undersample, columns=['Class'])
```

```python
fig, axs = plt.subplots(ncols=2, figsize=(13, 4.5))
sns.countplot(x='Class', data=data_validated, ax=axs[0])
sns.countplot(x='Class', data=test, ax=axs[1])

fig.suptitle('Class repartition before and after undersampling')
a1 = fig.axes[0]
a1.set_title('Before')
a2 = fig.axes[1]
a2.set_title('After')
```

## **Feature Engineering**
#### menambahkan informasi baru (kolom baru) yang akan digunakan untuk training model

```python
# 1. buat fitur/kolom Amount_log

# melakukan log transformation pada Amount
data_validated['Amount_log'] = np.log1p(
    data_validated['Amount']
)

# bandingkan skewness sebelum dan sesudah transformasi
print("Skewness Original Amount:",
      data_validated['Amount'].skew())

print("Skewness Log Amount:",
      data_validated['Amount_log'].skew())
```

```python
fig, axs = plt.subplots(ncols=2, figsize=(13, 5))

# sebelum transformasi log (Amount)
sns.histplot(data_validated['Amount'], bins=50, kde=True, color='blue', ax=axs[0])
axs[0].set_title('Before Log Transformation (Amount)')
axs[0].set_xlabel('Amount')

# sesudah transformasi log (Amount_log)
sns.histplot(data_validated['Amount_log'], bins=50, kde=True, color='orange', ax=axs[1])
axs[1].set_title('After Log Transformation (Amount_log)')
axs[1].set_xlabel('Amount_log')

fig.suptitle('Amount Distribution Before and After Log Transformation')
plt.tight_layout()
plt.show()
```

```python
# 2. buat fitur/kolom Hour
# mengekstraksi informasi waktu dalam siklus 24 jam karena sebelumnya "Time" masih dalam detik
data_validated['Hour'] = ((data_validated['Time'] % 86400) // 3600).astype(int)

# 3. buat fitur/kolom Hour_sin
# mengubah fitur Hour menjadi representasi siklik (agar berdekatan)
data_validated['Hour_sin'] = np.sin(2 * np.pi * data_validated['Hour'] / 24)

# 3. buat fitur/kolom Hour_cos
# mengubah fitur Hour menjadi representasi siklik (agar berdekatan)
data_validated['Hour_cos'] = np.cos(2 * np.pi * data_validated['Hour'] / 24)

# menghapus fitur Hour sementara
data_validated.drop(columns=['Hour'], inplace=True)

# Melihat hasil feature engineering
data_validated[['Time', 'Hour_sin', 'Hour_cos']].head()
```

```python
# check setelah feature engineering dan preprocessing
# melihat seluruh fitur
print("Daftar Kolom:")
print(data_validated.columns)

print("\nUkuran Dataset:")
print(data_validated.shape)

# periksa missing values 
print("\nMissing Values:")
print(data_validated.isnull().sum().sum())

# check distribusi label
print("\nDistribusi Class:")
print(data_validated['Class'].value_counts())

# check hasil akhir dataset
data_validated.head()
```
