### 3. Uniqueness Validation

```python
duplicate_data = data.duplicated()

if duplicate_data.sum() > 0: 
    print("Ada duplikat")
    print("Total duplikat : ", duplicate_data.sum())
    duplicate_data = data[data.duplicated(keep=False)]
    display(duplicate_data)
else: 
    print("Tidak ada duplikat")
```

```python
duplicate_count = data.duplicated().sum()
duplicate_percentage = (duplicate_count / len(data)) * 100

print(f"Duplicate Count: {duplicate_count}")
print(f"Duplicate Percentage: {duplicate_percentage:.4f}%")
```

```python
print("Original Dataset:")
print(data['Class'].value_counts())

print("\nDuplicate Records:")
print(data[data.duplicated()]['Class'].value_counts())
```

```python
duplicate_groups = data.groupby(
    list(data.columns),
    dropna = False
).size()

duplicate_groups = duplicate_groups[duplicate_groups > 1]

print('Number of Duplicate Groups:', len(duplicate_groups))
print('Total Rows in Duplicate Groups:', duplicate_groups.sum())
print('Extra Duplicate Rows:', (duplicate_groups - 1).sum())

print('\nDuplicate Group Sizes:')
print(duplicate_groups.value_counts().sort_index())
```

### Undersampling (untuk balancing, mengurangi jumlah sampel kelas mayoritas)

```python
import imblearn
from imblearn.under_sampling import RandomUnderSampler

undersample = RandomUnderSampler(sampling_strategy=0.5)
```

```python
X = data.drop(columns=['Class'])
Y = data['Class']

# undersample
X_undersample, Y_undersample = undersample.fit_resample(X, Y)
```

```python
test = pd.DataFrame(Y_undersample, columns=['Class'])
```

```python
fig, axs = plt.subplots(ncols=2, figsize=(13, 4.5))
sns.countplot(x='Class', data=data, ax=axs[0])
sns.countplot(x='Class', data=test, ax=axs[1])

fig.suptitle('Class repartition before and after undersampling')
a1 = fig.axes[0]
a1.set_title('Before')
a2 = fig.axes[1]
a2.set_title('After')
```
